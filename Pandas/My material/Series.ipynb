#Series in Pandas
A series is a 1-D labelled array capable of holding any data type. This axis lables are collectively called the index.
Series are vertical but vectors are horizontal

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a','b','c']
my_list = [10,20,30]
arr = np.array([1,2,3])
d = {1:10,2:20,3:30}
```

```python
pd.Series(my_list)
```

```python
pd.Series(my_list,index = labels)
```

```python
pd.Series(arr)
```

```python
pd.Series(d)
```
