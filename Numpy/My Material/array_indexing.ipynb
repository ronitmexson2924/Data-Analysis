```python
arr[0,0]
```

```python
arr[0,4]
```

#BOOLEAN INDEXING

```python
arr = np.arange(11,21)
arr
```

```python
bool_index = arr % 2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
