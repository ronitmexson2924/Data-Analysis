#Array generation functions


```python
#arange function
arr1 = np.arange(1,11)
arr1
```

```python
arr2 = np.arange(1,11,2)
arr2
```

```python
#generate zeros
arr3 = np.zeros(5)
arr3
```

```python
arr4 = np.zeros((4,8))
arr4
```

```python
#generate ones
arr5 = np.ones(6)
arr5
```

```python
arr6 = np.ones((8,4))
arr6
```

```python
#linspace - Linear Spacing
#this is used to generate the elements in an array such that they have linear spaces between them.
arr7 = np.linspace(1,6,2)
arr7
```

```python
arr8 = np.linspace(1,5,10)
arr8
```

```python
arr9 = np.linspace(0,1,100)
arr9
```
