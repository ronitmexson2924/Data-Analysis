#Array Operations

```python
import numpy as np
```

**Arithmetic operations**

```python
a1 = np.array([1,2,3,4,5])
a2 = np.array([6,7,8,9,10])
```

```python
arr2 = a1 ** a2
arr2
```

**Broadcasting**

```python
l = [10,20,30,40]
arr = np.array(l)
arr
```

```python
arr+10
```

```python
arr2 = np.arange(1,26).reshape(5,5)
arr2
```

```python
arr2 = arr2 + 10
arr2
```

**Deep and shallow copy**

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
slice = slice * 10
slice
```

```python
a
```

```python
a[4:6] * 10
```

```python
a
```

```python
a
b = a #deep copy
```

```python
b[0] = 99
b
```

```python
a #changes made i copy were also reflected in the main array
```

```python
b = a.copy() #shallow copy
b
```

```python
b[6] = 55
b
```

```python
a #clearly we can see the difference
```

**Matrix Operations**

```python
mat1 = np.arange(1,5).reshape(2,2)
mat2 = np.arange(5,9).reshape(2,2)
mat1
```

```python
mat2
```

```python
#matrix multiplication
mat1 @ mat2
```

```python
#Transpose
mat1.T
```
